# Setup: Catalog and Unity Catalog structure
Creates the `olist` catalog and the schemas for each Medallion architecture layer (Bronze, Silver, Gold), plus the landing zone volume for raw files.

```python
%sql
CREATE CATALOG IF NOT EXISTS olist;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS olist.bronze;
CREATE SCHEMA IF NOT EXISTS olist.silver;
CREATE SCHEMA IF NOT EXISTS olist.gold;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS olist.bronze.raw_landing;
```

```python
%sql
SHOW SCHEMAS IN olist;
```
